### 2.1 A bowl contains 10 balls, 6 red and 4 green. Two balls are drawn at random, with replacement&mdash;any ball withdrawn is replaced before the next ball is drawn. Describe the sample space $S$ and the probability function $P$. How many points does the sample space contain?

### Answer
Because the experiment records the color on each of two draws with replacement, $S$ contains all ordered color outcomes:

$ S = \{(R, R), (R, G), (G, R), (G, G)\}$

As seen, the sample space (S) has 4 outcomes (points).


Since there are more red balls than green balls, outcomes with more red balls are more likely:

$ P(R, R) = \frac{6}{10} \cdot \frac{6}{10} = \frac{36}{100} = 0.36 = 36\% $

$ P(R, G) = \frac{6}{10} \cdot \frac{4}{10} = \frac{24}{100} = 0.24 = 24\% $

$ P(G, R) = \frac{4}{10} \cdot \frac{6}{10} = \frac{24}{100} = 0.24 = 24\% $

$ P(G, G) = \frac{4}{10} \cdot \frac{4}{10} = \frac{16}{100} = 0.16 = 16\% $

### 2.3 How many two-digit numbers can be formed from 1, 2, 3, 4, 5, if duplication of the integers is not allowed? If duplication is allowed?

### Answer

This is a simple combination and permutation question. For no duplications, but order mattering, we can use the Permutation formula:

$P(n, k) = \frac{n!}{(n-k)!}$

Where, $n = 5$ and $k = 2$. This thus gives:

$P(5, 2) = \frac{5!}{(5-2)!} = \frac{5 \times 4 \times 3!}{3!} = 20$

For duplication being allowed, we just need to add the remaining 5 possibilities: $11, 22, 33, 44, 55$. Making it $25$ numbers.


### 2.4 How many three-digit numbers can be formed from 0, 1, 2, 3, 4 if duplication is not allowed? If duplication is allowed?

### Answer

Again, Permutation problem. This time, though, $n = 5$ and $k = 3$ (this is assuming a number that starts with 0 is allowed):

$ P(5, 3) = \frac{5!}{(5-3)!} = \frac{5 \times 4 \times 3 \times 2!}{2!} = 60$

Now, for duplication being allowed, we need $P(5,3) + P(3,2) * C(5,2) + 5$, 

the $P(5,3)$ being no duplicates, $P(3, 2) * C(5, 2)$ being 2 repeated numbers and $5$ being the 3 repeated numbers

This yields: $125$



## 3. In how many ways can $n$ different objects be divided into $k$ groups containing $n_1, n_2, \ldots, n_k$ objects, if

$$
    n_1 + n_2 + \ldots + n_k = n-m,\ {\rm where}\ n > m?
$$

### Answer

There are $m$ objects left over after forming the $k$ groups.

Choose $n_1$ objects for the first group, then $n_2$ from the remaining objects, and so on:

$$
\binom{n}{n_1}
\binom{n-n_1}{n_2}
\binom{n-n_1-n_2}{n_3}
\cdots
\binom{m+n_k}{n_k}
$$

This simplifies to:

$$
\frac{n!}{n_1!n_2!\cdots n_k!m!}
$$

So, for $n$ distinct objects divided into $k$ groups of sizes
$n_1,n_2,\ldots,n_k$, with $m$ objects left over, the number of possible divisions is

$$
\frac{n!}{m!\prod_{i=1}^{k} n_i!}
$$


## 4 Let ${\bf x}_1$, ${\bf x}_2$ be a random sample of size 2 from a normal distribution with unknown mean $\mu$ and variance 1, where $\mu$ can vary between $-\infty$ and $+\infty$. We want to find a point estimate of $\mu$ considering three estimators:

$$
\begin{align}
    \hat\mu_1 & = d_1({\bf x}_1, {\bf x}_2) = {\bf x}_1\\
    \hat\mu_2 & = d_2({\bf x}_1, {\bf x}_2) = 1/4 {\bf x}_1 + 3/4 {\bf x}_2 \\

and the loss function

$$

### Answer

The risk of estimator $d$, when the true parameter is $\mu$, is the expected loss of using $d(X_1,X_2)$ to estimate $\mu$.

How much loss do we expect this estimator to have on average over many repeated experiments: 

$$
R(d, \mu) = E_\mu[L(d(X_1, X_2), \mu)]
$$

$ X_i = \mu + noise $, where noise has mean 0 and variance 1.

$$
E[g(x)] = \int_{-\infty}^{\infty} g(x)f_X(x)dx
$$

For a Normal Distribution, $ f_X(x) = \frac{1}{\sigma\sqrt{2\pi}} e^{-\frac{(x-\mu)^2}{2\sigma^2}} $, where variance $\sigma^2 = 1$ and, from the loss function: $ g(x) = 2 \mu (x-\mu)^2 $

So:

$$
E[2\mu(X - \mu)^2] = 2\mu E[(X-\mu)^2] = \int_{-\infty}^{\infty} 2\mu(x-\mu)^2 \frac{1}{\sqrt{2\pi}} e^{-\frac{(x-\mu)^2}{2}} dx =  2\mu \int_{-\infty}^{\infty} (x-\mu)^2 \frac{1}{\sqrt{2\pi}} e^{-\frac{(x-\mu)^2}{2}} dx = 2\mu
$$

$$
E[(X - \mu)^2] = 1 = E[(X - E[X])^2] = E[X^2] - E[X]^2 = Var(X)
$$

Also:

$$
E[(X-\mu)^2] = Var(X) + Bias(X)^2, Bias(X) = E[X] - \mu
$$

# Programming Question - Outlier Detection

Our goal today is to build a tool to detect outliers using basic statistics. The solution is going to be very simple, but despite this, it is actually in line with how a lot of manufacturing engineers would traditionally approach this kind of problem (think "six sigma" people). Usually we turn to machine learning and data science because the problems we are interested in are not this trivial. Hopefully, this exercice will introduce you into programing probabilistically, and serve to motivate why more complicated algorthims are needed in the future.

Imagine you are a manufacturing engineer. Every month you ship out 1000 units of product to your customer. The customer complains that 1% of all units failed prematurely. You are tasked with identifying these defective units before they ship out so that customers don't get defective parts. To test your method you will be working on historical data, where we already know which parts customers returned as defective (labels variable is 0 for good 1 for bad) Let's explore one way this problem could be solved.  

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn import datasets

n_samples = 1000
dimensionality = 2  # number of features (in this case, sensors)
outlier_prev = 0.01  # previous i.e. prior probability of outlier
data, label = datasets.make_blobs(
    n_samples=[int(n_samples * (1 - outlier_prev)), int(n_samples * outlier_prev)],
    centers=[[0] * dimensionality, [5] + [0] * (dimensionality - 1)],
    random_state=32,
)
# n_samples is the total number of samples, and the first argument to make_blobs is a list of the
# number of samples for each cluster. The first cluster has 99% of the samples, and the second
# cluster has 1% of the samples (outliers).
# The centers argument specifies the center of each cluster.
# The first cluster is centered at (0,0), and the second cluster is centered at (5,0).
# The random_state argument is used to seed the random number generator for reproducibility.

trans = [[0.6, -0.6], [-0.4, 0.8]]  # Transformation matrix
data = np.dot(data, trans)  # Apply the transformation to the data to make it more interesting

plt.title("Sensor Data")
plt.xlabel("Sensor 1")
plt.ylabel("Sensor 2")
plt.scatter(data[:, 0], data[:, 1])
plt.show()
```

Inspecting the dataset above we can see that it consists of data from 2 sensors taken during the manufacturing process. Each point corresponds to a single product that was manufactured. We also see that the data seems to be clumped around a center point with the density of points decreasing as we move away from the center of the clump. Luckily, we have a language to describe the centrality and the spread of the data quantitatively - the mean and the variance.

Compute the mean and the variance of the data below. Note, since the data contains 2 features the covariance will be a matrix, and the mean a vector.

```python
# Compute the mean of the data along each feature (column)
mean = np.mean(data, axis=0)  # Not 0,0 as outlier cluster is at (5,0) and skews the mean

# Compute the covariance matrix of the data;
cov = np.cov(data.T)  # .T transpose as np.cov expects features as rows and samples as columns

print(mean)
print(cov)
```

A good way to idetify outliers would be to ask how far a point is from the mean. Since we know that 99% of our product is good, it may be a good guess that the mean corresponds to a good product. While measuring the distance from the mean is good, how do we know what distance corresponds to a product being a defect? Furthermore, the distance is just part of the story, does direction also play a role?

All of these are key considerations, and there are many ways you could try to solve this. However, the language of probability density functions is a natural way to address all of these concerns. We are going to use a basic distribution here, which is what makes this method so limited and full of assumptions. However, the idea of a probability distribution is very flexible, and research on how to model complex probability distributions is at the core of many modern advancements in machine learning. (you may have heard that chatGPT predicts the "probability" of the next word based on the previous ones).

So what is a probability distribution? They come in two main flavors - conditional and unconditional. For now lets focus on the unconditional case. Simply put an unconditional probability distribution is just a function that returns how likely an event is to occur. You can imagine for a coin toss the function would look something like this.


$$
P(X = x) =
\begin{cases}
\frac{1}{2}, & \text{if } x \in \{ \text{Heads}, \text{Tails} \} \\
0, & \text{otherwise}
\end{cases}
$$

Here, X  represents the outcome of the coin flip, and x  can be either "Heads" or "Tails". Each outcome has a probability of $\frac{1}{2}$, and any other outcome has a probability of 0.

When dealing with probability distributions we can either fit them or evaluate them. Fitting is deciding what the function should be. In the case of the coin flip it would be deciding the chance of landing heads or tails (assuming the coin may be unfair). This is usually extracted from data (something like observing the coin was heads 100 times and tails 3). Evaluating is given a fit function how likely is an event (asking what the chance of getting a head is given someone else already fit the function). For the coin example this is all trivial, but things get more complicated when the data is continous.

For continous variables we no longer deal with probabilities but instead probability densities, the act of fitting becomes known as density estimation. Here is a video on why this is: https://www.youtube.com/watch?v=Fvi9A_tEmXQ.

For this problem we will use the Multivarate Gaussian Distribution shown below. This distribution takes in the mean and covarance we already computed (providing these "fits" the gaussian to the data). Note, the mean and covaraince are $\mu,\Sigma$ respectivly.

![](https://wikimedia.org/api/rest_v1/media/math/render/svg/418f7f0a42778c344e9bff0876adb348655cd367)

This distribution will tell us the probability density of a die coming from our dataset. Please code it below. Note, Make sure that it can take the whole data array in at once, not just a single points at a time.

```python
def multivariate_gaussian_density(data, mean, cov):
    difference = data - mean
    k = len(mean)

    # Each row becomes Σ⁻¹(xᵢ - μ)
    solved = np.linalg.solve(cov, difference.T).T

    # Get inside exp
    exponent = -0.5 * np.sum(difference * solved, axis=1)
    denominator = np.sqrt((2 * np.pi) ** k * np.linalg.det(cov))

    density = np.exp(exponent) / denominator
    return density


density = multivariate_gaussian_density(data, mean, cov)

plt.scatter(data[:, 0], data[:, 1], c=density)
plt.title("Predicted Probability Densities")
plt.xlabel("Sensor 1")
plt.ylabel("Sensor 2")
plt.colorbar(label="Gaussian Density")
plt.show()
```

Here we have the predicted probability density for each product. Its clear to see that the density is higher the closer you are to the center, but also the density decays at different rates in different directions.

Now that we have this additional density information lets use it to predict what points are outliers. To do this write a code snippet that flags the 1% least likely points as outliers (since we know customers are reporting about a 1% failure rate).

### Detecting Outliers

```python
# simple outlier detector
threshold = np.quantile(density, 0.01)
label_pred = (density <= threshold).astype(int)  # Array of 1s and 0s to color plot

fig, ax = plt.subplots(1, 2, figsize=(15, 5), dpi=200)
ax[0].set_title("True Outliers")
ax[0].scatter(data[:, 0], data[:, 1], c=label)
ax[1].set_title("Detected Outliers")
ax[1].scatter(data[:, 0], data[:, 1], c=label_pred)
plt.show()
```

Here we can see that the true outliers (that we know from customer feedback) closely align with the outliers our procedure detected. The classification is not perfect, but it's much better than ignoring the problem all togther! Let's quantify this performance rather than just looking at it. Below we will plot the confusion matrix, and report the F1 Score. We will learn about these measures later in the class, but for now just know that in this case the higher the F1 score the better with 1 being perfect.

You can read about them here:

https://en.wikipedia.org/wiki/Confusion_matrix

https://en.wikipedia.org/wiki/F-score

```python
from sklearn.metrics import ConfusionMatrixDisplay, f1_score

ConfusionMatrixDisplay.from_predictions(label, label_pred)
f1 = f1_score(label, label_pred)
print(f"The F1 score is {f1}")
```
