## 1. Variables, types, operators, and strings — 4 points

### Values and types

- `int`: a whole number, such as `3`
- `float`: a decimal number, such as `1.25`
- `str`: text inside quotes, such as `"hello"`
- `bool`: either `True` or `False`
- `None`: the absence of a value

### Assignment and operators

- `=` assigns a value to a name.
- `==` compares two values for equality.
- Python does not require a type declaration before a variable name.
- Arithmetic operators include `+`, `-`, `*`, `/`, `//`, `%`, and `**`.
- Comparison operators such as `<`, `<=`, `==`, and `!=` produce booleans.
- An **f-string** begins with `f` and inserts expressions inside `{}`.

### Functions in this project

- Every TODO is a function: inputs arrive as parameters and the result leaves via `return`.
- `return` hands a value back to the caller; `print` only displays it.
- A function that ends without an executed `return` produces `None`.

```python
count = 3
distance = 6.0
speed = 2.5
name = "Ada"
is_moving = True
print(type(count), type(distance), type(name), type(is_moving))
time = distance / speed
message = f"{name} needs {time} seconds"
print(message)
print("Is time positive?", time > 0)
```

### Common mistakes

- Using `=` when you mean to compare with `==`
- Forgetting quotation marks around a string
- Assuming `print(value)` returns the value
- Mixing up `/` with `//`, which discards the fractional part

## 2. Conditionals and loops — 8 points

An **algorithm** is a precise sequence of steps.

### Conditionals

- `if` tests the first condition.
- `elif` tests another condition when earlier branches were false.
- `else` handles all remaining cases.
- `and`, `or`, and `not` combine or modify boolean expressions.
- Python uses indentation to identify the code inside each branch.
- `raise ValueError('message')` inside an `if` rejects an invalid input; Section 4 covers errors in detail.

### Loops

- A `for` loop visits every item in a collection.
- An **accumulator** stores a running result, such as a count or total.
- A `while` loop repeats while its condition remains true.
- A `while` loop must change something that eventually makes its condition false.

```python
temperature = 14
if temperature < 10:
    label = "cold"
elif temperature < 25:
    label = "mild"
else:
    label = "hot"
print(label)

below_count = 0
for value in [0.2, 1.0, 0.9, 3.0]:
    if value < 1.0:
        below_count += 1
print(below_count)

remaining, steps = 25, 0
while remaining > 0:
    remaining -= 10
    steps += 1
print(steps)
```

### Common mistakes

- Using the wrong boundary: `< 10` excludes 10, while `<= 10` includes it
- Forgetting to initialize an accumulator before the loop
- Creating an infinite `while` loop by never changing its condition
- Starting a loop before validating its inputs

### Practice 2 — 8 points

## 3. Lists, tuples, and dictionaries — 8 points

### Lists

- A list is ordered and mutable: `[1, 2, 3]`.
- Indexing starts at zero; `values[-1]` selects the final item.
- `values[start:stop]` returns a slice and excludes `stop`.
- `.append(item)` adds an item to the end.
- Useful built-ins include `len`, `sum`, `min`, and `max`.

### Tuples

- A tuple is ordered and immutable: `(x, y)`.
- `x, y = position` unpacks its items into separate variables.
- Tuples are useful for fixed records such as coordinates.

### Dictionaries

- A dictionary maps unique keys to values.
- `record['name']` retrieves a value by key.
- `record.get('team', 'none')` provides a default when a key is absent.
- Dictionaries are mutable.

```python
values = [4, 1, 7]
print(values[0], values[-1], values[1:])
values.append(2)
print("length, sum, min, max:", len(values), sum(values), min(values), max(values))

position = (2.0, 3.0)
x, y = position
print(x, y)

record = {"name": "Ada", "score": 85, "position": [0, 0]}
print(record["name"])
print(record.get("team", "none"))
record["score"] = 90
```

### Mutation, copying, and edge cases

- `alias = values` does not copy a list; both names refer to the same object.
- Use `values.copy()` or `values[:]` to create a separate shallow copy.
- Dictionaries also mutate in place.
- `min([])` is undefined and raises an error.
- `sum([]) / len([])` divides by zero, so reject empty input first.

## 4. Useful Python iteration patterns — 8 points

- A **list comprehension** builds a list: `[expression for item in values]`.
- An optional `if condition` filters items in a comprehension.
- `range(n)` represents integers from 0 through n−1.
- `enumerate(values)` produces `(index, value)` pairs.
- `zip(a, b)` pairs corresponding items from two iterables.
- `zip` stops when the shorter iterable ends.
- `item in collection` checks membership.
- Prefer readable comprehensions over complicated one-line expressions.

```python
numbers = [1, 2, 3]
squares = [number**2 for number in numbers]
print(squares)
for index, value in enumerate(numbers):
    print(index, value)
names, prices = ["apple", "banana"], [1.2, 0.7]
for name, price in zip(names, prices, strict=False):
    print(name, price)
print(list(range(4)))
```

### Common mistakes

- Expecting `range(4)` to include 4
- Forgetting that `zip` stops at the shorter collection
- Forgetting that `enumerate` starts at zero by default

## 5. Packages and NumPy basics — 12 points

### Modules and packages

- A **module** is a file containing reusable Python code.
- A **package** groups related modules.
- `import numpy as np` imports NumPy using its conventional alias.

### NumPy arrays

- Arrays store numeric data efficiently.
- Operations such as multiplication apply elementwise.
- `[1, 2] * 2` repeats a Python list.
- `np.array([1, 2]) * 2` computes `[2, 4]`.
- `shape` describes an array's dimensions.
- A matrix with 2 rows and 3 columns has shape `(2, 3)`.
- `axis=0` combines rows and returns one result per column.
- `axis=1` combines columns and returns one result per row.
- `np.linalg.norm` computes Euclidean length.

```python
print([1, 2] * 2)
array = np.array([1, 2], dtype=float)
print(array * 2, array.shape, array.dtype)
matrix = np.array([[1, 2], [3, 6]], dtype=float)
print("shape:", matrix.shape)
print("first row:", matrix[0])
print("first column:", matrix[:, 0])
print("column means:", np.mean(matrix, axis=0))
print("length:", np.linalg.norm(np.array([3, 4])))
```

### Common mistakes

- Expecting Python lists to support NumPy arithmetic
- Ignoring an unexpected array shape
- Choosing the wrong axis for a reduction
- Modifying the caller's input unnecessarily

## 6. Classes, objects, attributes, and modules — 8 points

### Classes and objects

- A **class** defines a kind of object.
- `__init__` initializes a new instance.
- A default such as `x=0` makes a constructor argument optional.
- `self` refers to the current object.
- `self.x` is an attribute that stores persistent object state.
- Calling `point.move(...)` passes `point` as `self` automatically.
- Methods may mutate attributes, so later calls see updated values.

### The `math` module

- `math` belongs to Python's standard library.
- `math.sin` and `math.cos` expect angles in radians.
- Use `math.radians(degrees)` before applying trigonometric functions.

```python
class ExampleCounter:
    def __init__(self, start=0):
        self.value = start

    def increment(self):
        self.value += 1
        return self.value


counter = ExampleCounter(5)
print(counter.increment(), counter.increment())  # state persists: 6, then 7
```

### Common mistakes

- Forgetting `self` in a method signature
- Writing `x` instead of `self.x` and losing persistent object state
- Passing degrees directly to `sin` or `cos`
